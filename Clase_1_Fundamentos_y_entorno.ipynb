## 1. Definiciones importantes

Antes de escribir una sola línea de código, necesitamos entender **cinco
conceptos** que vamos a usar durante todo el resto del año:

**Programa · Lenguaje · Algoritmo · Pseudocódigo · Diagrama de flujo**

Tomémonos esta primera parte con calma: si estos conceptos quedan claros,
todo lo que viene después se entiende mucho más fácil.

### Programa

> **Definición.** Un **programa** es un conjunto de instrucciones, escritas en
> algún lenguaje de programación, que ejecutadas **secuencialmente** (una después
> de la otra) resuelven un problema específico.

**En palabras simples:** un programa es una **lista de órdenes** que la
computadora va a obedecer **en orden**, de arriba hacia abajo.

La computadora no "piensa" ni "imagina" lo que queremos: hace **exactamente** lo
que le decimos, en el **orden exacto** en que se lo decimos.

### Programa — un ejemplo cotidiano

Pensá en las instrucciones para **preparar un mate**:

1. Calentar el agua (sin que hierva)
2. Llenar el mate con yerba
3. Inclinar el mate y poner la bombilla
4. Cebar el agua
5. Tomar

Eso es un "programa" para una persona: pasos **concretos**, en **orden**.
Un programa de computadora es lo mismo, pero escrito en un lenguaje que la
máquina entiende.

### Lenguaje

> **Definición.** Un **lenguaje** es una serie de símbolos que sirven para
> transmitir uno o más mensajes (ideas) entre **dos entidades diferentes**.

El español es un lenguaje entre personas. El **lenguaje de programación** es el
que usamos para comunicarnos con la **computadora**.

### Lenguaje — ¿por qué uno especial?

Las personas nos entendemos aunque hablemos de forma ambigua o incompleta
("alcanzame *eso* de *ahí*"). La computadora **no** tolera ambigüedad: necesita
instrucciones exactas.

Por eso existen los **lenguajes de programación**: tienen reglas estrictas
(la **sintaxis**) que no dejan lugar a dudas. En esta materia usaremos **Python**.

> Existen muchos lenguajes (Python, C, Java, JavaScript...). Cada uno tiene sus
> reglas, igual que el español y el inglés son dos idiomas distintos.

### Algoritmo

> **Definición.** Un **algoritmo** es una serie de pasos organizados que
> describen el proceso que se debe seguir para dar solución a un problema
> específico.

**Importante:** el algoritmo es **la idea**, la estrategia para resolver el
problema. Es **independiente** del lenguaje: el *mismo* algoritmo se puede
escribir después en Python, en C o en cualquier otro lenguaje.

> Primero pensamos el **algoritmo** (cómo resolverlo) y recién después
> escribimos el **programa** (en un lenguaje concreto).

### Algoritmo — la analogía de la receta

Una **receta de cocina** es un algoritmo:

- Tiene **ingredientes** (los datos de entrada).
- Tiene **pasos en orden** (las instrucciones).
- Produce un **resultado** (el plato terminado).

Si dos personas siguen la misma receta con los mismos ingredientes, obtienen
el mismo plato. Un algoritmo funciona igual.

### Algoritmo — propiedades que SIEMPRE debe cumplir

| Propiedad | Significado | Ejemplo de qué pasa si falla |
|-----------|-------------|------------------------------|
| **Preciso** | No se presta a interpretaciones ambiguas | "Agregar *un poco* de sal" → ¿cuánto es un poco? |
| **Definido** | Con los mismos datos, da siempre el mismo resultado | Que a veces dé 90 y a veces 91 |
| **Finito** | Tiene comienzo y fin: un número determinado de pasos | Un paso que dice "seguir para siempre" |

### Algoritmo — ejemplo paso a paso

**Problema:** decidir si una persona puede votar (en Argentina, desde los 16).

```
Inicio
    Leer la edad de la persona
    Si la edad es mayor o igual a 16:
        Mostrar "Puede votar"
    Si no:
        Mostrar "Todavía no puede votar"
Fin
```

Es **preciso** (la condición es clara), **definido** (misma edad → misma
respuesta) y **finito** (termina siempre).

### Pseudocódigo

> **Definición.** El **pseudocódigo** es una mezcla entre un lenguaje de
> programación y el idioma de un país (en nuestro caso, el español). Se usa
> para **diseñar** un algoritmo antes de programarlo.

El prefijo *pseudo-* significa "falso": es un código **"falso"**, que **no se
ejecuta** en ninguna computadora. Sirve para que **nosotros** pensemos y
ordenemos la lógica con palabras, sin preocuparnos todavía por la sintaxis
estricta de un lenguaje.

### Pseudocódigo — ejemplo

```
Inicio
    Leer L1
    Leer L2
    Area <- L1 * L2
    Mostrar Area
Fin
```

- `<-` significa "asignar" (guardar un valor en una variable).
- No hay reglas estrictas: lo importante es que **se entienda la lógica**.

> El pseudocódigo es como el **borrador** de un escrito: primero ordenamos las
> ideas y después las pasamos en limpio (al lenguaje de programación).

### Diagrama de flujo

> **Definición.** Un **diagrama de flujo** es la representación **gráfica** de
> un algoritmo. Usa distintos tipos de figuras unidas por flechas que indican
> la **secuencia** del proceso.

Mientras el pseudocódigo describe el algoritmo con **palabras**, el diagrama de
flujo lo describe con un **dibujo**. Las flechas muestran el camino que sigue
la ejecución, paso a paso.

### Diagrama de flujo — los símbolos básicos

| Figura | Nombre | Para qué se usa |
|--------|--------|-----------------|
| Óvalo | Inicio / Fin | Marca dónde empieza y termina |
| Rectángulo | Proceso | Un cálculo o una acción (ej.: `Area = L1 * L2`) |
| Paralelogramo | Entrada / Salida | Leer datos o mostrar resultados |
| Rombo | Decisión | Una pregunta con respuesta Sí/No |
| Flecha | Flujo | Indica el orden en que se avanza |

En la próxima sección los vamos a ver dibujados con un ejemplo real.

### Diagrama de flujo — un ejemplo con decisión

Para decidir si una persona **puede votar** (edad ≥ 16) usamos un **rombo** de decisión, del que salen dos caminos: **Sí** y **No**.

Diagrama de flujo con decisión: ¿puede votar?

### Un algoritmo se puede expresar de varias formas

Ya tenemos los conceptos. Ahora: el **mismo** algoritmo se puede escribir de
**cuatro** maneras distintas.

1. Con una **fórmula matemática**
2. Con un **diagrama de flujo** (gráfico)
3. Con **pseudocódigo** (lenguaje natural estructurado)
4. Con un **lenguaje de programación** (Python, C, Java...)

Vamos a verlas todas con el **mismo ejemplo**.

## 2. Un mismo problema, cuatro formas

> **Problema:** calcular la superficie de un rectángulo a partir de sus dos lados.

### a) Como fórmula matemática

Definimos las variables:

- Lados: $L_1$ y $L_2$
- Resultado: $Area$

$$Area = L_1 \times L_2$$

Con $L_1 = 5$ y $L_2 = 18$ → $Area = 90$

### b) Como diagrama de flujo

Diagrama de flujo: superficie de un rectángulo

Las figuras siguen los símbolos que vimos recién: **óvalo** (inicio/fin), **paralelogramo** (leer/mostrar) y **rectángulo** (proceso).

### c) Como pseudocódigo

```
Inicio
    Leer L1
    Leer L2
    Area <- L1 * L2
    Mostrar Area
Fin
```

El pseudocódigo es independiente del lenguaje: describe **la lógica**, no la sintaxis.

### d) Como programa en Python

Fijate lo **parecido** que es al pseudocódigo 👇

````python
L1 = 5
L2 = 18
Area = L1 * L2
print(Area)
````

Esa cercanía entre **pseudocódigo** y **Python** es una de las razones por las que
vamos a aprender a programar con este lenguaje.

## 3. ¿Por qué Python?

### Compilador vs. intérprete

Para que la computadora entienda nuestro código hay que **traducirlo** a lenguaje
de máquina (binario). Hay dos estrategias:

- **Compilador** (ej.: C): traduce **todo** el programa antes de ejecutarlo.
- **Intérprete** (ej.: **Python**): traduce y ejecuta **instrucción por instrucción**.

Python es **interpretado**: escribimos una línea, la corremos y vemos el resultado
al instante. Ideal para aprender.

### Ventajas de Python para nosotros

- Sintaxis **clara y cercana al pseudocódigo**
- **No** hay que declarar tipos ni reservar memoria a mano
- Resultado **inmediato** (intérprete)
- Enorme uso **profesional** en ingeniería: análisis de datos, simulación,
  optimización, automatización
- Gratis y multiplataforma

## 4. Nuestro entorno de trabajo

### ¿Qué es un IDE (entorno de desarrollo)?

> **Definición.** Un **IDE** (del inglés *Integrated Development Environment*,
> "entorno de desarrollo integrado") es un programa que **reúne en un solo lugar**
> todas las herramientas necesarias para escribir, ejecutar y corregir programas.

Se llama "integrado" porque **junta** varias herramientas que de otro modo
estarían separadas. Las tres principales son:

- **Editor de código** — para *escribir* el programa.
- **Intérprete / compilador** — para *traducirlo* y *ejecutarlo*.
- **Depurador** (*debugger*) — para *encontrar y corregir* errores.

### Las herramientas de un IDE

**Editor de código.** Es donde escribimos el **código fuente**: el conjunto de
instrucciones, escritas en el lenguaje de programación (Python), que forman
nuestro programa.

**Intérprete (o compilador).** Cumple dos funciones: (1) **revisa** que el
código esté bien escrito y nos avisa si hay un **error de sintaxis**, y
(2) lo **traduce** a lenguaje de máquina para ejecutarlo. Como vimos, Python
usa un **intérprete** (ejecuta línea por línea).

**Depurador (*debugger*).** Su función es ayudarnos a **encontrar y reducir los
errores** (*bugs*) de un programa.

> *Nota:* en lenguajes **compilados** como C existe además un **vinculador**
> (*linker*), que arma el archivo ejecutable (`.exe`). En Python, al ser
> **interpretado**, ese paso **no existe**.
